```python

board = [' ' for _ in range(9)]
HUMAN = 'X'
COMPUTER = 'O'


def display_board():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()


def check_win(player):
    winning_combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_combinations:
        if (board[a] == player and
                board[b] == player and
                board[c] == player):
            return True

    return False


def check_draw():
    return ' ' not in board


def minimax(is_maximizing):
    if check_win(COMPUTER):
        return 10

    if check_win(HUMAN):
        return -10

    if check_draw():
        return 0

    if is_maximizing:
        best_score = -float('inf')

        for i in range(9):
            if board[i] == ' ':
                board[i] = COMPUTER
                score = minimax(False)
                board[i] = ' '
                best_score = max(best_score, score)

        return best_score

    else:
        best_score = float('inf')

        for i in range(9):
            if board[i] == ' ':
                board[i] = HUMAN
                score = minimax(True)
                board[i] = ' '
                best_score = min(best_score, score)

        return best_score


def find_best_move():
    best_score = -float('inf')
    best_move = None

    for i in range(9):
        if board[i] == ' ':
            board[i] = COMPUTER
            score = minimax(False)
            board[i] = ' '

            if score > best_score:
                best_score = score
                best_move = i

    return best_move


def human_move():
    while True:
        try:
            position = int(input("Enter your position (1-9): "))

            if position < 1 or position > 9:
                print("Please enter a number from 1 to 9.")

            elif board[position - 1] != ' ':
                print("That position is already occupied.")

            else:
                board[position - 1] = HUMAN
                break

        except ValueError:
            print("Please enter a valid number.")


def play_game():
    global board
    board = [' ' for _ in range(9)]

    print("================================")
    print("       TIC-TAC-TOE GAME")
    print("================================")
    print("You are X")
    print("Computer is O")
    print()

    print("Board positions are:")
    print()
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")
    print("\nGame starts!\n")

    while True:
        display_board()
        print("Your turn.")
        human_move()

        if check_win(HUMAN):
            display_board()
            print("Congratulations! You WIN!")
            break

        if check_draw():
            display_board()
            print("The game is a DRAW!")
            break

        print("Computer is thinking...")
        best_move = find_best_move()
        board[best_move] = COMPUTER

        if check_win(COMPUTER):
            display_board()
            print("Computer WINS!")
            break

        if check_draw():
            display_board()
            print("The game is a DRAW!")
            break


if __name__ == "__main__":
    play_game()
```
