```python

def dls(state, goal, depth, path):
    if state == goal:
        return path + [state]

    if depth == 0:
        return None

    zero = state.index(0)
    moves = [-3, 3, -1, 1]  # Up, Down, Left, Right

    for move in moves:
        new_pos = zero + move

        if new_pos < 0 or new_pos >= 9:
            continue
        if move == -1 and zero % 3 == 0:
            continue
        if move == 1 and zero % 3 == 2:
            continue

        new_state = list(state)
        new_state[zero], new_state[new_pos] = (
            new_state[new_pos], new_state[zero]
        )
        new_state = tuple(new_state)

        if new_state not in path:
            result = dls(new_state, goal, depth - 1,
                         path + [state])
            if result:
                return result

    return None


def iddfs(start, goal):
    depth = 0

    while True:
        result = dls(start, goal, depth, [])
        if result:
            return result
        depth += 1


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

result = iddfs(start, goal)

print("Solution found!")
print("Steps:", len(result) - 1)

for state in result:
    print(state[:3])
    print(state[3:6])
    print(state[6:])
    print()
```
