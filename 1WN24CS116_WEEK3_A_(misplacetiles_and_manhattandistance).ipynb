```python
import heapq

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# h(n) = number of misplaced tiles
def heuristic(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Generate possible next states
def children(state):
    result = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            result.append(tuple(new_state))

    return result


# A* algorithm
def astar(start):

    # priority queue: (f, g, state, path)
    pq = []

    h = heuristic(start)

    heapq.heappush(pq, (h, 0, start, [start]))

    visited = set()

    while pq:

        f, g, state, path = heapq.heappop(pq)

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)

        for next_state in children(state):

            if next_state not in visited:

                new_g = g + 1
                new_h = heuristic(next_state)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, next_state, path + [next_state])
                )





    return None


# Initial state
start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)


solution = astar(start)

# Print solution
for state in solution:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```

```python
import heapq

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


def heuristic(state):
    distance = 0

    for i in range(9):
        tile = state[i]

        # Ignore the blank tile
        if tile != 0:
            # Current position
            current_row = i // 3
            current_col = i % 3

            # Goal position
            goal_index = goal.index(tile)
            goal_row = goal_index // 3
            goal_col = goal_index % 3

            # Manhattan distance
            distance += abs(current_row - goal_row) + abs(current_col - goal_col)

    return distance

# Generate possible next states
def children(state):
    result = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            result.append(tuple(new_state))

    return result


# A* algorithm
def astar(start):

    # priority queue: (f, g, state, path)
    pq = []

    h = heuristic(start)

    heapq.heappush(pq, (h, 0, start, [start]))

    visited = set()

    while pq:

        f, g, state, path = heapq.heappop(pq)

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)

        for next_state in children(state):

            if next_state not in visited:

                new_g = g + 1
                new_h = heuristic(next_state)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, next_state, path + [next_state])
                )





    return None


# Initial state
start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)


solution = astar(start)

# Print solution
for state in solution:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```
