```python

import random

def heuristic(board, n):
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def hill_climbing(n):
    while True:
        board = [random.randint(0, n - 1) for _ in range(n)]

        while True:
            current_h = heuristic(board, n)

            if current_h == 0:
                return board

            best_board = board[:]
            best_h = current_h

            for col in range(n):
                for row in range(n):
                    if board[col] != row:
                        new_board = board[:]
                        new_board[col] = row
                        h = heuristic(new_board, n)

                        if h < best_h:
                            best_h = h
                            best_board = new_board

            if best_h >= current_h:
                break

            board = best_board


def display(board, n):
    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


n = int(input("Enter the value of N: "))

if n < 4:
    print("No solution exists for this N.")
else:
    solution = hill_climbing(n)
    print("Solution found:")
    display(solution, n)
    print("Heuristic value:", heuristic(solution, n))
```
