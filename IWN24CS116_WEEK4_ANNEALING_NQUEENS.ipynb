```python

import random
import math

def heuristic(board):
    conflicts = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing():
    while True:
        board = list(range(8))
        random.shuffle(board)

        temperature = 10.0

        while temperature > 0.01:
            h = heuristic(board)

            if h == 0:
                return board

            col = random.randint(0, 7)
            new_row = random.randint(0, 7)

            new_board = board[:]
            new_board[col] = new_row

            new_h = heuristic(new_board)
            delta = h - new_h

            if delta > 0 or random.random() < math.exp(delta / temperature):
                board = new_board

            temperature *= 0.95


def display(board):
    for row in range(8):
        for col in range(8):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


solution = simulated_annealing()

print("Solution found:")
display(solution)
print("Heuristic value:", heuristic(solution))
```
