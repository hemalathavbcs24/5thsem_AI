```python

facts = {
    "Barks(Tommy)",
    "FourLegs(Tommy)",
    "SharpTeeth(Tommy)"
}

rules = [
    (["Barks(Tommy)", "FourLegs(Tommy)"], "Dog(Tommy)"),
    (["Feathers(Tommy)", "LaysEggs(Tommy)"], "Bird(Tommy)"),
    (["Dog(Tommy)"], "Mammal(Tommy)"),
    (["Mammal(Tommy)", "SharpTeeth(Tommy)"], "Carnivore(Tommy)"),
    (["Bird(Tommy)"], "CanFly(Tommy)")
]

def forward_chaining(facts, rules, goal):

    while True:
        new_fact_found = False

        for conditions, conclusion in rules:

            if all(condition in facts for condition in conditions):

                if conclusion not in facts:
                    facts.add(conclusion)
                    print("Derived:", conclusion)
                    new_fact_found = True

                if goal in facts:
                    return True

        if not new_fact_found:
            break

    return goal in facts


goal = "Carnivore(Tommy)"

if forward_chaining(facts, rules, goal):
    print("\nGoal proved: Tommy is a Carnivore")
else:
    print("\nGoal cannot be proved")
```
